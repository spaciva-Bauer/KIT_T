## Teil 1 – Warum testen wir Programme?

Ein Programm, das keine Fehlermeldung erzeugt, ist **nicht automatisch korrekt**.
Logische Fehler laufen ohne Absturz durch – sie liefern einfach falsche Ergebnisse.

Ein **Testfall** besteht aus:
- **Eingabe** – der Wert, der eingegeben wird
- **Erwartete Ausgabe** – was das *korrekte* Programm liefern soll
- **Tatsächliche Ausgabe** – was das Programm wirklich ausgibt
- **Bewertung** – bestanden (✓) oder nicht bestanden (✗)

> **Wichtig:** Die erwartete Ausgabe wird **vor** dem Test festgelegt – nicht erst nach dem Ausführen!

---
## Teil 2 – Arten von Testfällen

| Typ | Abk. | Beschreibung | Beispiel (Punkte 0–15) |
|-----|------|--------------|------------------------|
| Normalfall | NF | Typische Eingabe aus dem Kernbereich | `punkte = 11` |
| Grenzfall | GF | Genau an einer Bereichsgrenze | `punkte = 13`, `12`, `0`, `15` |
| Fehlerfall | FF | Außerhalb des gültigen Bereichs | `punkte = -1`, `16` |

**Äquivalenzklassen:** Eingaben, die zum selben Verhalten führen, bilden eine Klasse.
Es genügt, *einen* Vertreter pro Klasse zu testen.

Für das Notenprogramm (FOS/BOS Bayern, Punkte 0–15):

| Klasse | Bereich | Ausgabe |
|--------|---------|--------|
| Ungültig | < 0 | `Ungültige Eingabe` |
| Ungenügend | 0 | `ungenügend` |
| Mangelhaft | 1–3 | `mangelhaft` |
| Ausreichend | 4–6 | `ausreichend` |
| Befriedigend | 7–9 | `befriedigend` |
| Gut | 10–12 | `gut` |
| Sehr gut | 13–15 | `sehr gut` |
| Ungültig | > 15 | `Ungültige Eingabe` |

## Teil 3 – Das Notenprogramm mit Testfalltabelle

Unten siehst du das vollständige Notenprogramm und die zugehörige Testfalltabelle.

| Nr. | Beschreibung | Eingabe | Erwartete Ausgabe | Typ |
|-----|--------------|---------|-------------------|-----|
| T01 | Normalfall: sehr gut | 14 | `sehr gut` | NF |
| T02 | Grenzfall: untere Grenze sehr gut | 13 | `sehr gut` | GF |
| T03 | Grenzfall: obere Grenze gut | 12 | `gut` | GF |
| T04 | Normalfall: befriedigend | 8 | `befriedigend` | NF |
| T05 | Grenzfall: untere Grenze gut | 10 | `gut` | GF |
| T06 | Grenzfall: 0 Punkte | 0 | `ungenügend` | GF |
| T07 | Grenzfall: obere Grenze gültig | 15 | `sehr gut` | GF |
| T08 | Fehlerfall: negative Punktzahl | -1 | `Ungültige Eingabe` | FF |
| T09 | Fehlerfall: Wert über 15 | 16 | `Ungültige Eingabe` | FF |

```python
# Das Notenprogramm (FOS/BOS Bayern: Punkte 0–15)
punkte = int(input('Punkte (0–15): '))

if   punkte >= 13 and punkte <= 15: print('sehr gut')
elif punkte >= 10:                  print('gut')
elif punkte >= 7:                   print('befriedigend')
elif punkte >= 4:                   print('ausreichend')
elif punkte >= 1:                   print('mangelhaft')
elif punkte == 0:                   print('ungenügend')
else:                               print('Ungültige Eingabe')
```
