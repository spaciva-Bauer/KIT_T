## Teil 1 – Variablen und Zuweisung

Eine **Variable** ist ein benannter Speicherplatz. Mit `=` wird ihr ein Wert zugewiesen – das ist **kein** mathematisches Gleichheitszeichen, sondern ein *Zuweisungsoperator*.

```
variablenname = Wert
```

````python
# Variablen anlegen und Werte zuweisen
vorname = 'Lena'
alter   = 18
groesse = 1.72
aktiv   = True

# Wert überschreiben – der alte Wert geht verloren
alter = 19

print(vorname, alter, groesse, aktiv)
````

## Teil 2 – Die vier Grunddatentypen

| Datentyp | Bedeutung | Beispielwerte |
|----------|-----------|---------------|
| `int`    | Ganze Zahl | `0`, `42`, `-7` |
| `float`  | Dezimalzahl | `3.14`, `-0.5`, `1.0` |
| `str`    | Text (Zeichenkette) | `'Hallo'`, `'Python'` |
| `bool`   | Wahrheitswert | `True`, `False` |

Mit `type()` kannst du den Datentyp einer Variablen abfragen.

````python
# type() zeigt den Datentyp an
x = 42
print(type(x))      # <class 'int'>

pi = 3.14159
print(type(pi))     # <class 'float'>

gruss = 'Hallo!'
print(type(gruss))  # <class 'str'>

flag = True
print(type(flag))   # <class 'bool'>
````

## Teil 3 – Operatoren

**Arithmetische Operatoren:** `+` `-` `*` `/` `//` `%` `**`

> **Wichtig:** `/` liefert immer `float`. `//` ist die Ganzzahldivision. `%` gibt den Rest zurück.

**Vergleichsoperatoren:** `==` `!=` `<` `>` `<=` `>=`  
Das Ergebnis ist immer `True` oder `False`.

````python
# / liefert immer float, // schneidet den Rest ab
print(7 / 2)    # 3.5
print(7 // 2)   # 3
print(7 % 2)    # 1  (Rest)
print(2 ** 8)   # 256

# + bei Strings bedeutet Verkettung, nicht Addition
print('Hallo' + ' Welt')  # Hallo Welt
print('Ha' * 3)           # HaHaHa
````
