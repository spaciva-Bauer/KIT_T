## Teil 1 – Ausgabe mit `print()`

```
print(*objekte, sep=' ', end='\n')
```

- **`sep`** – Trennzeichen zwischen mehreren Werten (Standard: Leerzeichen)
- **`end`** – Zeichen am Ende der Ausgabe (Standard: Zeilenumbruch)

> **Komma vs. Plus:** Mit Komma können Werte verschiedener Typen direkt ausgegeben werden. Mit `+` müssen alle Teile Strings sein.

````python
# Einfache Ausgaben
print('Hallo Welt')              # Hallo Welt
print(42)                        # 42
print(3.14, True)                # 3.14 True

# Mehrere Werte mit sep trennen
print('a', 'b', 'c', sep='-')   # a-b-c

# Ohne Zeilenumbruch am Ende
print('Ergebnis:', end=' ')
print(100)                       # Ergebnis: 100

# Variablen ausgeben
name  = 'Lena'
alter = 18
print('Name:', name, '| Alter:', alter)  # Name: Lena | Alter: 18
````

## Teil 2 – Eingabe mit `input()`

```
input(aufforderung='')
```

`input()` hält das Programm an und wartet auf eine Tastatureingabe.  
**Das Ergebnis ist immer ein `str`** – auch wenn der Benutzer eine Zahl eingibt!

> Für Rechenoperationen muss der Rückgabewert mit `int()` oder `float()` umgewandelt werden.

````python
# Eingabe als String speichern
name  = input('Dein Name: ')

# Eingabe als Zahl: Typumwandlung notwendig!
alter = int(input('Dein Alter: '))
preis = float(input('Preis in Euro: '))

# Verarbeitung und Ausgabe
print('Hallo', name + '!')
print('In 10 Jahren bist du', alter + 10, 'Jahre alt.')
````

## Teil 3 – Kommentare und das EVA-Prinzip

Kommentare beginnen mit `#` und werden von Python **nicht ausgeführt**.

Ein gut strukturiertes Programm folgt dem **EVA-Prinzip**: erst **E**ingabe, dann **V**erarbeitung, dann **A**usgabe.

````python
# Programm: Flächenberechnung eines Rechtecks (EVA-Prinzip)

# --- Eingabe ---
laenge = float(input('Länge in m: '))   # Eingabe der Länge
breite = float(input('Breite in m: '))  # Eingabe der Breite

# --- Verarbeitung ---
flaeche = laenge * breite

# --- Ausgabe ---
print('Fläche:', flaeche, 'm²')
````
