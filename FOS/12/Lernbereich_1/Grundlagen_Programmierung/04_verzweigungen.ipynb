## Teil 1 – Einfache Verzweigung: `if` / `else`

Eine Verzweigung führt einen Codeblock nur aus, wenn eine **Bedingung wahr** (`True`) ist.

```
if <Bedingung>:
    <Anweisungen wenn True>
else:
    <Anweisungen wenn False>
```

> **Einrückung ist Pflicht!** Alle Zeilen innerhalb eines `if`/`else`-Blocks müssen um **4 Leerzeichen** eingerückt sein. Fehlende Einrückung führt zu einem `IndentationError`.

````python
temperatur = int(input('Temperatur in °C: '))

if temperatur > 35:
    print('Hitzealarm! Bitte viel trinken.')
else:
    print('Temperatur im normalen Bereich.')
````

---
## Teil 2 – Mehrfache Verzweigung: `if` / `elif` / `else`

Mit `elif` (kurz für ‘else if’) können **mehrere Bedingungen** nacheinander geprüft werden.
Sobald eine zutrifft, werden alle weiteren **übersprungen**.

```
if <Bedingung 1>:
    <Anweisungen>
elif <Bedingung 2>:
    <Anweisungen>
else:
    <Anweisungen (kein Fall passte)>
```

> **Reihenfolge zählt!** Python prüft von oben nach unten. Die engste Bedingung muss zuerst kommen.

````python
punkte = int(input('Punkte (0–100): '))

if punkte >= 90:
    print('Note: Sehr gut')
elif punkte >= 75:
    print('Note: Gut')
elif punkte >= 50:
    print('Note: Bestanden')
else:
    print('Note: Nicht bestanden')
````

---
## Teil 3 – Logische Operatoren

Bedingungen können mit **logischen Operatoren** verknüpft werden:

| Operator | Bedeutung | Beispiel | Ergebnis |
|----------|-----------|----------|----------|
| `and` | Beide müssen wahr sein | `x > 0 and x < 10` | `True` wenn x in {1…9} |
| `or` | Mindestens eine muss wahr sein | `x < 0 or x > 100` | `True` wenn x < 0 oder > 100 |
| `not` | Kehrt den Wahrheitswert um | `not x == 0` | `True` wenn x ungleich 0 |

````python
# Ampelsteuerung mit logischen Operatoren
farbe = input('Ampelfarbe (rot/gelb/grün): ')

if farbe == 'rot' or farbe == 'gelb':
    print('Bitte warten!')
elif farbe == 'grün':
    print('Fahren erlaubt.')
else:
    print('Unbekannte Farbe.')

# and statt verschachteltem if
alter = int(input('Alter: '))
if alter >= 18 and alter <= 67:
    print('Erwerbtätig')
else:
    print('Nicht erwerbtätig')
````

### Übung 2 – `elif`-Zweig ergänzen

Das Temperaturprogramm kennt bisher nur Hitzealarm (> 35 °C) und den normalen Bereich.

**Aufgabe:** Ergänze einen `elif`-Zweig für Frost (< 0 °C) mit der Meldung `Vorsicht Frost!`.
Der `else`-Zweig soll nur noch für den normalen Bereich (0–35 °C) gelten.

### Übung 3 – Syntaxfehler finden und beheben

**Aufgabe:** Das Programm enthält einen klassischen Syntaxfehler. Führe es aus, lies die Fehlermeldung und behebe ihn.

> **Tipp:** In einer Bedingung wird `==` (Vergleich) verwendet, **nicht** `=` (Zuweisung).

````python
# Finde und behebe den Syntaxfehler
passwort = input('Passwort eingeben: ')

if passwort = 'geheim123':     # ← Fehler!
    print('Zugang gewährt!')
else:
    print('Falsches Passwort.')
````
