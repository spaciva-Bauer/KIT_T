## Teil 1 – Die `for`-Schleife

Eine `for`-Schleife wiederholt einen Codeblock für **jeden Wert in einer Folge**.

```python
for <Laufvariable> in <Folge>:
    <Anweisungen>
```

Die Funktion `range()` erzeugt Zahlenfolgen. **Der Stopwert ist ausgeschlossen.**

| Aufruf | Erzeugte Folge |
|--------|----------------|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(2, 10, 2)` | 2, 4, 6, 8 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

> **Akkumulator-Muster:** Variable vor der Schleife initialisieren (z. B. `summe = 0`) und in jedem Schritt aktualisieren.

````python
# Einmaleins einer Zahl
n = int(input('Welche Zahl? '))

for i in range(1, 11):
    print(n, 'x', i, '=', n * i)
````

````python
# Summe der Zahlen 1 bis 100 (Gaußsche Summe)
summe = 0                       # Akkumulator initialisieren

for i in range(1, 101):
    summe = summe + i           # in jedem Schritt addieren

print('Summe 1 bis 100:', summe)  # 5050
````

## Teil 2 – Die `while`-Schleife

Eine `while`-Schleife wiederholt einen Block, **solange eine Bedingung wahr ist**. Sie wird eingesetzt, wenn die Anzahl der Durchläufe vorher nicht bekannt ist.

```python
while <Bedingung>:
    <Anweisungen>
```

> **Endlosschleife vermeiden!** Die Laufvariable muss sich innerhalb der Schleife verändern.

````python
# Countdown mit while
i = 5
while i > 0:
    print(i)
    i = i - 1       # i muss sich verändern!
print('Start!')
````

## Teil 3 – `break` und `continue`

- **`break`** verlässt die Schleife sofort
- **`continue`** überspringt den Rest des aktuellen Durchlaufs und geht zum nächsten

Beide funktionieren in `for`- und `while`-Schleifen.

| Wann welche Schleife? |
|----------------------|
| **`for`**: Anzahl der Wiederholungen ist bekannt (Einmaleins, Summe bis n) |
| **`while`**: Abbruchbedingung hängt von Laufzeit-Werten ab (Benutzereingabe) |

````python
# break: Schleife verlassen, sobald eine negative Zahl eingegeben wird
for i in range(1, 6):
    z = int(input('Zahl eingeben: '))
    if z < 0:
        print('Negative Zahl gefunden – Abbruch!')
        break
````

````python
# continue: gerade Zahlen überspringen, nur ungerade ausgeben
for i in range(1, 11):
    if i % 2 == 0:
        continue    # gerade Zahlen überspringen
    print(i)        # Ausgabe: 1 3 5 7 9
````

### Übung 2 – Off-by-One-Fehler beheben

**Aufgabe:** Das Programm soll die Summe aller Zahlen von 1 bis einschließlich `n` berechnen. Für `n = 5` sollte die Ausgabe `15` sein, es erscheint aber `10`. Finde und behebe den Fehler.

````python
# Finde den Fehler – das Ergebnis für n=5 muss 15 sein, nicht 10!
n = int(input('Bis zu welcher Zahl? '))
summe = 0

for i in range(1, n):     # ← hier liegt der Fehler
    summe = summe + i

print('Summe 1 bis', n, '=', summe)
````

### Übung 3 – Akkumulator: Produkt berechnen

**Aufgabe:** Das folgende Programm summiert Zahlen (Akkumulator). Ändere es so, dass es stattdessen das **Produkt** aller Zahlen von 1 bis `n` berechnet (Fakultät: n!).

> Tipp: Das Produkt startet mit `1` (nicht `0`). Statt `+` brauchst du `*`.

````python
# Ändere den Akkumulator: Summe → Produkt (Fakultät n!)
n = int(input('n = '))

summe = 0              # ← Startwert anpassen

for i in range(1, n + 1):
    summe = summe + i  # ← Operation anpassen

print(n, '! =', summe)
# Erwartung für n=5: 120  (5! = 1*2*3*4*5)
````
