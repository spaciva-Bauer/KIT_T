## Teil 3: Textformatierung

Mit Markdown kannst du Text auf verschiedene Weisen hervorheben:

| Syntax | Ergebnis |
|---|---|
| `**Text**` | **fett** |
| `*Text*` | *kursiv* |
| `~~Text~~` | ~~durchgestrichen~~ |
| `` `Text` `` | `Code im Fließtext` |

- das Wort `print` als Codetext formatiert ist

Führe danach die Zelle aus, um das Ergebnis zu sehen.

## Teil 7: Mathematische Formeln

Jupyter unterstützt **LaTeX-Formeln** direkt in Markdown-Zellen:

- **Inline** (im Fließtext): `$Formel$` → der Satz des Pythagoras lautet $a^2 + b^2 = c^2$
- **Abgesetzt** (eigene Zeile, zentriert): `$$Formel$$`

$$E = mc^2$$

Häufig verwendete LaTeX-Symbole:

| Symbol | LaTeX-Code |
|---|---|
| $x^2$ | `x^2` |
| $x_i$ | `x_i` |
| $\\frac{a}{b}$ | `\\frac{a}{b}` |
| $\\sqrt{x}$ | `\\sqrt{x}` |

### Uebung 7.1 – Formel einbetten

Doppelklicke auf diese Zelle, um zu sehen wie Formeln eingebettet werden. Bearbeite dann die Zelle unten:

Die Fläche eines Kreises berechnet sich mit der Formel HIER INLINE-FORMEL EINFÜGEN.
