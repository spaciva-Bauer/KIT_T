#### Die Lernenden erleben, wie zwei spezialisierte Sprachmodelle als Pipeline zusammenarbeiten: Ein Bildbeschreibungsmodell erzeugt englischen Text aus einem Bild – ein Übersetzungsmodell überträgt diesen anschließend ins Deutsche.

## Ziel: Niederschwellige Darstellung einer Bild-zu-Text-Pipeline (Image Captioning)

> Unter **Image Captioning** versteht man die automatische Erzeugung einer natürlichsprachigen Bildbeschreibung durch ein KI-Modell. Das Modell analysiert den Bildinhalt und gibt einen beschreibenden Satz aus – ähnlich wie ein Mensch, der ein Bild in Worte fasst. Solche Systeme werden z.B. eingesetzt, um Bilder für sehbehinderte Menschen zugänglich zu machen, Bilddatenbanken automatisch zu verschlagworten oder Inhalte in sozialen Netzwerken zu moderieren.

## Die Pipeline im Überblick

Das System arbeitet in **zwei Phasen**, die du in diesem Notebook schrittweise umsetzt:

| Phase | Bezeichnung | Was passiert? |
|---|---|---|
| 1 | **Beschreibung** | Ein multimodales Modell analysiert das Bild und erzeugt eine englische Bildbeschreibung. |
| 2 | **Übersetzung** | Ein Übersetzungsmodell überträgt die englische Beschreibung ins Deutsche. |

---

```python
from transformers import pipeline  # Import der Funktion pipeline des Moduls transformers
from PIL import Image              # Import der Funktion Image des Moduls PIL (Pillow) zur Bildverarbeitung
```

### Phase 1 – Beschreibung
#### Zuerst wird das Bild geladen. Gib den Dateinamen deines Bildes an (das Bild muss im gleichen Ordner wie dieses Notebook liegen).

```python
# Trage hier den Dateinamen deines Bildes ein (z.B. "hund.jpg" oder "stadtpark.png").
# Das Bild muss im gleichen Ordner wie dieses Notebook gespeichert sein.
bilddatei = "Screenshot 2026-03-26 111402.png"

bild = Image.open(bilddatei).convert("RGB")  # Das Bild wird geöffnet und in das RGB-Format umgewandelt.
bild  # Zeigt das Bild direkt im Notebook an.
```

> **Aufgabe 1:** Lege ein eigenes Bild (Foto vom Smartphone, Screenshot o.ä.) im Ordner des Notebooks ab und trage den Dateinamen oben ein. Führe die Zelle danach aus. Das Bild wird im Notebook angezeigt.

#### Das Modell analysiert jetzt das Bild und erzeugt eine englische Bildbeschreibung.

Verwendet wird `Salesforce/blip-image-captioning-base` – ein multimodales Modell, das sowohl Bild- als auch Sprachverarbeitung beherrscht.

```python
# Das Modell wird geladen. Beim ersten Aufruf wird es automatisch von HuggingFace heruntergeladen.
# device=-1 verwendet die CPU (funktioniert auf jedem Rechner).
captioner = pipeline("image-to-text", model="Salesforce/blip-image-captioning-base", device=-1)

ergebnis = captioner(bild)  # Das Bild wird an das Modell übergeben.

beschreibung_englisch = ergebnis[0]["generated_text"]  # Die erzeugte Bildbeschreibung wird gespeichert.
print("Bildbeschreibung (Englisch): " + beschreibung_englisch)
```

> **Aufgabe 2:** Vergleiche die englische Bildbeschreibung mit dem tatsächlichen Bildinhalt. Was beschreibt das Modell korrekt? Was fehlt oder ist falsch? Notiere deine Beobachtungen.

---
### Phase 2 – Übersetzung
#### Die englische Bildbeschreibung wird nun von einem zweiten, spezialisierten Modell ins Deutsche übersetzt.

Verwendet wird `Helsinki-NLP/opus-mt-en-de` – ein Modell, das ausschließlich für die Übersetzung von Englisch nach Deutsch trainiert wurde.

```python
# Das Übersetzungsmodell wird geladen.
uebersetzer = pipeline("translation_en_to_de", model="Helsinki-NLP/opus-mt-en-de", device=-1)

ergebnis_uebersetzung = uebersetzer(beschreibung_englisch)  # Die englische Beschreibung wird übersetzt.

beschreibung_deutsch = ergebnis_uebersetzung[0]["translation_text"]  # Das Ergebnis wird gespeichert.
print("Bildbeschreibung (Deutsch): " + beschreibung_deutsch)
```

#### Zur besseren Übersicht werden beide Ergebnisse abschließend zusammen ausgegeben:

```python
print("=" * 50)
print("Ergebnis der Pipeline")
print("=" * 50)
print("Englisch : " + beschreibung_englisch)
print("Deutsch  : " + beschreibung_deutsch)
print("=" * 50)
```
