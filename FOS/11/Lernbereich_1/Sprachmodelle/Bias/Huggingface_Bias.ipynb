## 3. Das Modell: BERT und die Fill-Mask-Aufgabe

### Was ist BERT?

**BERT** (*Bidirectional Encoder Representations from Transformers*) ist ein Sprachmodell von Google (2018). Es wurde auf Milliarden von Texten trainiert und hat dabei gelernt, Sprache zu "verstehen" – also Zusammenhänge zwischen Wörtern zu erkennen.

### Was bedeutet "Fill-Mask"?

Bei der **Fill-Mask**-Aufgabe wird ein Wort in einem Satz durch den Platzhalter `[MASK]` ersetzt. Das Modell soll dann vorhersagen, welches Wort ursprünglich an dieser Stelle stand – basierend auf allem, was es beim Training gelernt hat.

**Beispiel:**
```
Eingabe:  "The capital of France is [MASK]."
Ausgabe:  "Paris"
```

Das Modell gibt dabei nicht nur einen Vorschlag, sondern die **5 wahrscheinlichsten Wörter** zurück – sortiert nach Wahrscheinlichkeit (das erste ist am wahrscheinlichsten).

````python
# Modell laden: bert-base-uncased (Englisch, Groß-/Kleinschreibung wird ignoriert)
unmasker = pipeline("fill-mask", model="bert-base-uncased")

# Satz 1: Mann in einem Beruf
result = unmasker("This man works as a [MASK].")
print("Mann:", [r["token_str"] for r in result])

# Satz 2: Frau in einem Beruf – identischer Satz, nur das Geschlecht geändert
result = unmasker("This woman works as a [MASK].")
print("Frau:", [r["token_str"] for r in result])
````

## 4. Ergebnisse lesen und verstehen

Die Ausgabe ist eine Liste der **5 wahrscheinlichsten Wörter** für `[MASK]`, in absteigender Reihenfolge.

- Das **erste** Wort in der Liste hält das Modell für am wahrscheinlichsten.
- Die anderen Wörter folgen mit abnehmender Wahrscheinlichkeit.
