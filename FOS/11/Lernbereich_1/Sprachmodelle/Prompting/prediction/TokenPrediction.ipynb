## 3. Wie funktioniert das Modell?

Wir nutzen ein Modell namens **BERT** (auf Deutsch trainiert). Es hat Millionen von deutschen Texten gelesen und dabei gelernt, welche Wörter häufig zusammen vorkommen.

**Das Prinzip:** Wir schreiben einen Satz mit einer Lücke – markiert durch `[MASK]` – und fragen die KI: Welches Wort passt hier am wahrscheinlichsten?

Beispiel:
```
"Die Sonne scheint heute sehr [MASK]."
                                 ^
                          Was kommt hier?
```

Die KI antwortet nicht mit einem einzigen Wort, sondern mit einer **Rangliste** von Vorschlägen – jeder mit einer Wahrscheinlichkeit in Prozent.

> **Wichtig:** Das Modell wird beim ersten Start heruntergeladen (~400 MB). Das kann **1–2 Minuten** dauern – das ist normal, einfach warten!

````python
from transformers import pipeline
import matplotlib.pyplot as plt

# Modell laden – beim ersten Mal wird es heruntergeladen, danach geht es schneller
print("Modell wird geladen ...")
modell = pipeline("fill-mask", model="dbmdz/bert-base-german-cased")
print("Fertig! Das Modell ist bereit.")
````

## 4. Erster Test

Die folgende Hilfsfunktion `zeige_ergebnisse(satz)` macht zwei Dinge:
1. Sie gibt die 5 wahrscheinlichsten Wörter als **Text** aus.
2. Sie zeigt die Wahrscheinlichkeiten als **Balkendiagramm** – so lassen sich die Ergebnisse leichter vergleichen.

Führe die Zelle aus und beobachte, welches Wort die KI für das `[MASK]` vorschlägt.

````python
def zeige_ergebnisse(satz):
    """Zeigt KI-Vorhersagen als Text und Balkendiagramm."""
    ergebnisse = modell(satz)

    print(f"Satz: {satz}\n")
    for erg in ergebnisse:
        print(f"  Wort: {erg['token_str']:<15} | Wahrscheinlichkeit: {round(erg['score'] * 100, 2)}%")

    woerter = [erg['token_str'] for erg in ergebnisse]
    werte   = [round(erg['score'] * 100, 2) for erg in ergebnisse]

    plt.figure(figsize=(7, 3))
    plt.barh(woerter[::-1], werte[::-1], color='steelblue')
    plt.xlabel("Wahrscheinlichkeit (%)")
    plt.title(f"Vorhersagen f\u00fcr: '{satz}'")
    plt.tight_layout()
    plt.show()


# --- Erster Test ---
zeige_ergebnisse("Das Wetter heute ist sehr [MASK].")
````
