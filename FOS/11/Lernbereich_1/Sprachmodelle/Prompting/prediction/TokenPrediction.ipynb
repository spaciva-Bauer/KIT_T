## 3. Das Modell laden
Wir nutzen ein Modell namens **BERT**. Es wurde mit Millionen deutschen Texten trainiert und kann Lücken in Sätzen füllen.

```python
from transformers import pipeline

# Wir erstellen eine Pipeline für 'fill-mask' (Lückentext füllen)
befehl = pipeline("fill-mask", model="dbmdz/bert-base-german-cased")
```

## 4. Erster Test
Schreibe einen Satz und nutze das Wort `[MASK]` für die Lücke. Die KI wird uns die 5 wahrscheinlichsten Wörter anzeigen.

```python
satz = "Das Wetter heute ist sehr [MASK]."
ergebnisse = befehl(satz)

for erg in ergebnisse:
    print(f"Wort: {erg['token_str']} | Wahrscheinlichkeit: {round(erg['score']*100, 2)}%")
```
